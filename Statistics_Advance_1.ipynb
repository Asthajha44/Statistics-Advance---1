## Theoretical Questions:

Q1. What is a random variable in probability theory?

Ans. In probability theory, a random variable is a mathematical function that assigns a numerical value to each possible outcome of a random event or experiment.


Q2. What are the types of random variables?

Ans. It is categorised into two main types;


*   Discrete random variable; which takes on a countable number of distinct values.e.g., the number of heads in 5 coin flips, roll of a die.
*   Continuous random variable; which takes on an infinite number of possible values within a given range.e.g., height, weight, exact time.



Q3. What is the difference between discrete and continuous distributions?

Ans. The key difference between discrete and continuous distributions are:

1. Value type:

* Discrete distribution; associated with random variables that take countable distinct values. e.g.,(0, 1, 2)

* Continuous distribution; associated with random variables that can take any value within a continuous range or interval. e.g.,(any real number between o and 10)

2. Probability measurement:

* Discrete; uses a probability mass function (PMF) to assign an exact non-zero probability to specific individual values (*P(X=x))*.

* Continuous; uses a probaility density fucntion (PDF) where the probability of any single exact point is zero. Probabilities are measured over an interval (the area under the curve).

3. Mathematical operation:

* Discrete; probabilities are summed across outcomes, with the total sum eqauling 1.

* Continuous; probabilities are integrated over an interval, with the total area under the curve equaling 1.

examples: Discrete - Binomial distribution, Poisson distribution

  Continuous - Normal distribution, Uniform distribution.










Q4. What are probability distribution functions (PDF)?

Ans. A probability distribution function is a mathematical expression that describes the likelihood of different possible outcomes for a random variable.
The primary functions are:


1. Probability Mass function
2. Probability Density Function
3. Cumulative Distribution Function




Q5. How do cumulative distribution functions (CDF) differ from probability distribution functions (PDF)?

Ans. The core differnce lies in what they measure:

* Probability function (PMF/PDF): Give the likelihood of a specific outcome or an exact point (for discrete variables) or relative likelihood within a tiny interval (for continuous variables).

* Cumulative distribution function: Gives the accumulated total probability that a random variable is less than or equal to a specific value (*P(X<=x)).*



Q6. What is a discrete uniform distribution?

Ans. A discrete uniform distribution is a probability distribution where a finite number of values are all equally likely to occur.

1. Probability: if there are n possible outcomes, the probability of any single outcome is 1/n.
2. Example: Rolling a fair 6 sided die, where they fce (1 through 6) has an identical 1/6 chance of landing faceup.


Q7. What are the key properties of a Bernoulli distribution?

Ans. A bernoulli distribution models a single trial or experiment with exactly two possible outcomes (success/failure). Its key properties are;

* Single trial- It represents only one trial (unlike the binomial distribution, which repeats trials).
* Single parameter (p): Governed by parameter p, representing the probability of success.The probability of failure is 1-p.

* Expected value (Mean): E(X)=p
* Variance: Var(X)=p(1-p).




Q8. What is the binomial distribution, and how is it used in probability?

Ans. A binomial distribution is a discrete probability distribution that models the total number of successes in a fixed number of independent yes/no (bernoulli) trials, where each trial has the exact same probability of success (p).

How it is used: it is used to calculate the likelihood of a specific number of successes in repeated independent events.
e.g., determining the probability of getting a certain number of heads in 10 coin flips, finding the defect rate in a manufacturing batch, or calculating pass/fail rates in a test.

Q9. What is the Poisson distribution and where is it applied?

Ans. A poisson distribution is a discrete probability distribution that measures the probability of a given number of independent events occuring within a fixed interval of time or space. It is governed by a single parameter, lambda, which represents the average rate or expected number of occurences in that interval.

Common applications: It is used to model count based events such as;
* Number of phone calls received by a call center in an hour.
* Number of customer arrivals at a store within a specific time frame.



Q10. What is a continuous uniform distribution?

Ans. A continuous uniform distribution is a probability distribution where every value within a specific, continuous interval [a,b] has an equal likelihood of occuring.
* Probability Density Function (PDF): It is constant across the interval, defined as *f(x)=1/b-a for a<=x<=b*.
* Example: Waiting for a bus that arrives uniformly and randomly between 0 and 10 minutes, where every exact moment within that time frame is equally probable.



Q11. What are the characteristics of a normal distribution?

Ans. A normal distribution is a continuos probability distribution with the following key characterstics;
*  Bell-shaped and symmetric: It features a symmetric, bell-shaped curve centered around its peak.
* Equal Central Tendencies: The mean, median, and mode are all equal and located at the exact center.
* Two parameters: It is completely defined by just two values; the mean and the standard deviation.
* Empirical Rule (68-95-99.7% rule): Roughly 68% of the data falls within 1 standard deviation of the mean, 95% within 2, and 99.7% within 3.
* Asymptotic Tails: The tails extend infinitely in both directions, approaching but never actually touching the horizontal axis.







Q12. What is the standard normal distribution, and why is it important?

Ans. A standard normal distribution is a specific normal distribution where the mean is 0 and the standard deviation is 1.

It is important because;
* Universal Standard: It acts as a common reference point for all normal distribution, regardless of their original units or scale.
* Simplified Calculations: By converting any normal variable into a Z-score, you can use a single standard lookup table (Z-table) or calculator to easily find probabilities, percentiles, and critical values for any normal dataset.



Q13. What is the Central Limit Theorem (CLT), and why is it critical in statistics?

Ans. The Central Limit Theorem (CLT) states that the distribution of sample means approximates a normal distribution as the sample size gets larger, regardless of the shape of the original population's distribution.

It is critical because;

 Enables Statistical Inference: It allows us to use normal distribution tools (like confidence intervals and hypothesis testing) on data from any popultiaon type, even if the underlying data is skewed or non-normal.

 Universal Foundation: It forms the mathematical backbone of inferential statistics, bridging the gap between small samples and large populations.

Q14. How does the Central Limit Theorem relate to the normal distribution?

Ans. The Central Limit Theorem connects directly to the normal distribution by stating that the sampling distribution of the sample mean becomes approximately normally distributed as the sample size increases, regardless of the shape of the original population's distribution.

Q15. What is the application of Z statistics in hypothesis testing?

Ans. A Z-statistics measures how many standard deviations a sample statistic is away from a hypothesized population parameter.

**Primary Applications:**

* Hypothesis testing for means and proportions: Used to determine if the difference between a sample result and a hypothesized population value is statistically significant or just due to random chance.
* Condtion for use: Applied when the sample size is large and when the population standard deviation is known.
* Decision making: The calculated Z-score is compared against a critical threshold to decide whether to reject the null hypothesis.





Q16. How do you calculate a Z-score, and what does it represent?

Ans. To calculate a Z-score: Subtract the population mean from the data value (x), then divide the result by the standard deviation.
It represents: A Z-score measures how many standard deviations a specific data point is away from the mean. A positive Z-score means the value is above the mean, while a negative Z-score means it is below the mean.

Q17. What are point estimates and interval estimates in statistics

Ans. Point Estimate: A single numerical value used to approximate an unknown population parameter (e.g., using a sample mean to estimate the population mean).

Interval Estimate: A range of values calculated from a sample data that is likely to contain the true population parameter, typically expressed with a specific level of confidence (e.g., a 95% confidence interval).

Q18. What is the significance of confidence intervals in statistical analysis?

Ans. The significance of confidence intervals includes:

* Quantifies Uncertainty: Instead of relying on a single point estimate, it provides a range of plausible values for a population parameter, reflecting the precision of the estimate.
* Measure Reliability: Indicates how confident we are (e.g., 95% confident) that the true population value falls within the calculated range.
* Aids Hypothesis Testing: Helps determine if a hypothesized population value is statistically plausible by checking whether it falls inside or outside the interval.





Q19. What is the relationship between a Z-score and a confidence interval?

Ans. The Z-score directly determines the width (margin of error) of a confidence interval based on the chosen confidence level.

* Formula Connection: A confidence interval is calculated using the Z-score and standard error.
Confidence Interval = Point Estimate += (Z x Standard Error)
* Impact: A higher confidence level requires a larger Z-score (e.g., 1.96 for 95%), which increases the margin of error and results in a wider interval.



Q20. How are Z-scores used to compare different distributions?

Ans. Z-scores allow you to compare data points from completely different distributions by converting them onto a common, standardized scale (mean of 0, standard deviation of 1).

* Removes Scale Differences: By measuring how many standard deviations a value is from its own mean, you can directly compare performance across entirely different units or datasets (e.g., comparing a score on a math exam vs. a history exam with different averages and spreads).
* Relative Standing: A higher Z-score indicates a higher relative position within its respective distribution, regardless of the original scale.



Q21. What are the assumptions for applying the Central Limit Theorem?

Ans. The key assumptions for applying the Central Limit Theorem are:

* Independence: The sampled observations must be independent of each other (typically ensured by random sampling).
* Sample size: The sample size must be sufficiently large, generally n>=30, particularly if the underlying population distribution is skewed or non-normal.
* Finite Variance: The population must have a finite variance and mean so that extreme values don't dominate the sum or average.





Q22. What is the concept of expected value in a probability distribution?

Ans. The expected value (E(X)) is the theoretical long-run average or weighted mean of all possible outcomes in a aprobability distribution, where each outcome is weighted by its probability of occuring.

* It acts as the "center of mass" or balance point of a random variable.
* It provides a mathematical prediction of the average outcome over many repeated trials (e.g., expected winnings in a game or expected return on an investment).



Q23. How does a probability distribution relate to the expected outcome of a random variable?

Ans. A probability distribution maps out all possible outcomes of a random variable along with their respective likelihoods.

* The expected outcome is the weighted average of these possibilities, using the probabilities from the distribution as weights.
* The distribution defines the likelihood of every individual outcome, whereas the expected value condenses that entire distribution into a single, long-run average center point.

```python
#2. Generate a discrete uniform distribution using Python and plot the probability mass function (PMF)?

import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import randint

#In Scipy, randint(low,high) covers integers from low up to high -1
#Here low=1, and high=7 to represent the outcomes of a 6-sided die (1,2,3,4,5,6)
low, high =1,7
x=np.arange (low,high)

#calculate the probability mass function (PMF)
#For a uniform distribution, each outcome has an equal probability: 1/ (high-low)
pmf_values=randint.pmf(x, low, high)
plt.figure(figsize=(8,5))
plt.stem(x, pmf_values, linefmt= 'b-', markerfmt= 'bo', basefmt= 'r-')
plt.title('Probability Mass Function of a Discrete Uniform Distribution', fontsize=12)
plt.xlabel('Possible Outcomes', fontsize=10)
plt.ylabel('Probability', fontsize=10)
plt.xticks(x)
plt.ylim(0, 0.25)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
#9. Write a Python function to calculate Z-scores from a dataset and plot them?

def plot_z_scores(data):
  z=(data-np.mean(data))/np.std(data)

  plt.figure(figsize=(8,5))
  plt.hist(z, bins=20, density=True, edgecolor='black')
  plt.axvline(x=0, color='red', linestyle='--', linewidth=1.5, label="Mean")
  plt.title("Histogram of Z-Scores")
  plt.xlabel("Z-Score")
  plt.ylabel("Frequency")
  plt.legend()
  plt.show()

data= np.random.normal(loc=0, scale=1, size=1000)
plot_z_scores(data)
```

```python
#16. Create a confidence interval for a dataset using Python and interpret the result?

from scipy.stats import t

data=[12,15,14,18,20,22,19,17,21,16]
mean, sem = np.mean(data), np.std(data,ddof=1)/np.sqrt(len(data))
ci=t.interval(0.95, df= len(data)-1, loc=mean, scale=sem)

print(f"Mean: {mean:.2f}")
print(f"95% Confidence Interval: ({ci[0]:.2f}, {ci[1]:.2f})")
```
